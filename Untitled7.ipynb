```python

def ordenar_numeros():
  """Lê três números inteiros e os imprime em ordem crescente."""

  numeros = []
  for i in range(3):
    while True:
      try:
        numero = int(input(f"Digite o {i+1}º número inteiro: "))
        numeros.append(numero)
        break
      except ValueError:
        print("Entrada inválida. Digite um número inteiro.")

  numeros.sort()
  print("Números em ordem crescente:", numeros)

ordenar_numeros()
```

```python

def is_prime(number):
  """Checks if a number is prime."""
  if number <= 1:
    return False
  if number <= 3:
    return True
  if number % 2 == 0 or number % 3 == 0:
    return False
  i = 5
  while i * i <= number:
    if number % i == 0 or number % (i + 2) == 0:
      return False
    i += 6
  return True

def main():
    """Gets user input and checks if it's a prime number."""
    try:
        num = int(input("Digite um número inteiro: "))
        if is_prime(num):
            print(f"{num} é um número primo.")
        else:
            print(f"{num} não é um número primo.")
    except ValueError:
        print("Entrada inválida. Por favor, digite um número inteiro.")

if __name__ == "__main__":
    main()
```

```python
def analyze_grades():
    """
    Lê as notas dos alunos, calcula a média e conta as notas acima/abaixo da média.
    """
    try:
        n = int(input())
        if not 0 < n <= 20000:
            print("Número de alunos inválido.")
            return

        grades = []
        for _ in range(n):
            try:
                grade = float(input())
                grades.append(grade)
            except KeyboardInterrupt:
                print("Entrada de dados interrompida pelo usuário.")
                return
            except ValueError:
                print("Entrada inválida. Certifique-se de inserir um número válido.")

        if not grades:
            print("Nenhuma nota inserida.")
            return

        average = sum(grades) / len(grades)

        above_average = 0
        below_average = 0

        for grade in grades:
            if grade > average * 1.10:
                above_average += 1
            elif grade < average * 0.90:
                below_average += 1

        print(f"{average:.2f}")
        print(above_average)
        print(below_average)

    except ValueError:
        print("Entrada inválida. Certifique-se de inserir números válidos.")

analyze_grades()
```

```python
def reverse_string():
  """Reads a string and prints its reversed content."""
  string = input()
  reversed_string = string[::-1]
  print(reversed_string)

reverse_string()
```

```python
def replace_characters():
    string = input()
    char1 = input()
    char2 = input()
    new_string = ""
    for char in string:
        if char == char1:
            new_string += char2
        else:
            new_string += char
    print(new_string)

replace_characters()
```

```python
def solve():
    x, y = map(int, input().split())

    if x > y:
        x, y = y, x

    for i in range(x, y + 1):
        if i % 5 == 2 or i % 5 == 3:
            print(i)

solve()
```

```python
def reverse_and_sort_numbers():
    """Reads integers, prints them in reverse and decreasing order."""
    numbers = []
    while True:
        try:
            line = input()
            if not line:
                break
            num = int(line)
            numbers.append(num)
        except ValueError:
            print("Invalid input. Please enter an integer.")
            continue

    print("Numbers in reverse order:")
    for num in reversed(numbers):
        print(num)

    numbers.sort(reverse=True)
    print("\nNumbers in decreasing order:")
    for num in numbers:
        print(num)

reverse_and_sort_numbers()
```
