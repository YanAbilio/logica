```python
# prompt: Faça um programa que leia 3 números inteiros e os imprima em ordem
# crescente

def ordenar_numeros():
  """Lê três números inteiros e os imprime em ordem crescente."""

  numeros = []
  for i in range(3):
    while True:
      try:
        numero = int(input(f"Digite o {i+1}º número inteiro: "))
        numeros.append(numero)
        break
      except ValueError:
        print("Entrada inválida. Digite um número inteiro.")

  numeros.sort()  # Ordena a lista em ordem crescente
  print("Números em ordem crescente:", numeros)

ordenar_numeros()
```
